Para la **Red 1** voy a utilizar los parámetros $ABCD$ definidos como:

$$V_1 = V_2 . A + (-I_2) . B  $$

$$I_1 = V_2 . C + (-I_2) . D  $$

Debido a la alta impedancia de salida del operacional podemos ver que ni $V_1$ ni $I_1$ dependen de la corriente de salida $I_2$ por lo que $B$ y $D$ son iguales a cero, quedando el resto de parámetros de la siguiente forma:

$$A = - \frac{R_1}{R_2}$$

$$B = 0$$

$$C = - \frac{1}{R_2}$$

$$D = 0$$

Para la **Red 2** también voy a utilizar los parámetros $ABCD$. Al tratarse de un transformador ideal Podemos ver que existe una relacion entre la tencion de entrada y salida:

$$\frac{V_1}{V_2} = \frac{N}{1} \rightarrow V_1 = N V_2$$

A su vez tambien existe una relación entre las corrientes de entrada y salida:

$$ I_1 = - \frac{I_2}{N} $$

Reemplazando en las ecuaciones de la definición:

$$V_1 = V_2 . N + (-I_2) . 0  $$

$$I_1 = V_2 . 0 + (-I_2) . \frac{1}{N}  $$

Por lo que los parámetros quedan de la siguiente forma:

$$A = N$$

$$B = 0$$

$$C = 0$$

$$D = \frac{1}{N}$$
